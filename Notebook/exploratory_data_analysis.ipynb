```python
import pandas as pd
import numpy as np
df = pd.read_csv(r"G:\My Drive\DATA Analysis\UCI Machine Learning Repository\diabetic_data_with_diagnosis_category.csv")

# total encounter
da= df.encounter_id.count()
print("Total_encounters =" ,da)

# unique patient 
db= df.patient_nbr.unique
print("Unique_patients =" ,db)

# 30-days readmission
dc= df.readmitted_30.sum
print(dc)

# 30-days readmission rate
dd = round(((df.readmitted_30.sum()) / df.encounter_id.count()) * 100, 2)
print(dd)

# Average length of stay
de= round(df.time_in_hospital.mean(), 2)
print(de)

# Average number of medications
dg = round( df["num_medications"].mean(),2)
print('average number of medications = ',dg)

# Average previous patient encounters
dh = round(df["number_inpatient"].mean(),2)
print(dh)
```

```python
# Readmission by rate

readmission_by_age = (

df.groupby("age", observed=False)
.agg(
    total_encounter = ("encounter_id","count"),
    readmitted_30 = ('readmitted_30',"sum"),
    readmission_rate = ("readmitted_30","mean")
)
.reset_index()
)
readmission_by_age["readmission_rate"] *= 100

print(readmission_by_age)
```

```python
# readmission rate by diagnosis group 

readmission_by_diagnosis_group = (

df.groupby("diag_1_category", observed=False)
.agg(
    total_encounter = ("encounter_id","count"),
    readmitted_30 = ('readmitted_30',"sum"),
    readmission_rate = ("readmitted_30","mean")
)
.reset_index()
)
readmission_by_diagnosis_group["readmission_rate"] = (readmission_by_diagnosis_group["readmission_rate"] * 100).round(2)

print(readmission_by_diagnosis_group)

```

```python
# Readmission rate by A1C result

readmission_by_A1C_result = (

df.groupby("A1Cresult", observed=False)
.agg(
    total_encounter = ("encounter_id","count"),
    readmitted_30 = ('readmitted_30',"sum"),
    readmission_rate = ("readmitted_30","mean")
)
.reset_index()
)
readmission_by_A1C_result["readmission_rate"] = (readmission_by_A1C_result["readmission_rate"] * 100).round(2)

print(readmission_by_A1C_result)

```
